PROBLEM STATEMENT

This project i about predicting how much time a Taxi ride will take to                                                                                                              
1.Reach the Location for pickup                                                                                                                                 
2.Time to Drop off to the Location                                                                                                
3.Keeping a track of time                                                                                             
4.And , to calculate number of pasenger using ML

DATA DICTIONARY

Column Name  - Description:         

id - Unique ID for each taxi trip                  
vendor_id   - ID of the taxi company                        
pickup_datetime  - Date and time when the trip started           
dropoff_datetime -  Date and time when the trip ended             
pickup_longitude -  Longitude of pickup location                  
pickup_latitude  - Latitude of pickup location                   
dropoff_longitude  - Longitude of drop-off location                
dropoff_latitude   -  Latitude of drop-off location                 
store_and_fwd_flag - Shows if data was stored before sending (Y/N)                                  
trip_duration   -  Total time of the trip in seconds
